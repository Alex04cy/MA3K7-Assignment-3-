

## Attack

I started by making both players choose random moves to see how the game worked. This helped me understand the rules, but it did not show who could force a win.

```python
import numpy as np
```

```python
def random_two_phase_nim(n, s):
    heaps = [0] * n
    player = 1
    # Phase 1: Placing stones
    print("Phase 1")
    for i in range(s):
        heap = np.random.randint(0, n)
        heaps[heap] += 1
        print(f"Player {player} places a stone in heap {heap+1}: {heaps}")
        player = 3 - player
    # Phase 2: Playing Nim
    print("\nPhase 2")
    while sum(heaps) > 0:
        heap = np.random.randint(0, n)
        if heaps[heap] > 0:
            take = np.random.randint(1, heaps[heap] + 1)
            heaps[heap] -= take
            print(f"Player {player} takes {take} stone(s): {heaps}")
            if sum(heaps) == 0:
                print(f"Player {player} loses.")
                print(f"Player {3-player} wins!")
                return
            player = 3 - player
random_two_phase_nim(6, 4)
```

### Looking for winning strategies

Next, I tried to make both players choose better moves, using recursion in a similar way to the Nim examples from the lectures. The function `winning` checks whether the player about to place a stone can force a win, and `clever` chooses a placement.

```python

# Check whether the next player can win at Nim

def nim_winning(heaps):
    if max(heaps) <= 1:
        return sum(heaps) % 2 == 0
    x = 0
    for h in heaps:
        x = x ^ h
    return x != 0
def winning(heaps, s):
    if s == 0:
        return nim_winning(heaps)
    for i in range(len(heaps)):
        new_heaps = heaps.copy()
        new_heaps[i] += 1
        if not winning(new_heaps, s-1):
            return True
    return False
def clever(heaps, s):
    for i in range(len(heaps)):
        new_heaps = heaps.copy()
        new_heaps[i] += 1
        if not winning(new_heaps, s-1):
            return i
    return 0
def two_phase_nim(n, s, play=False):
    print("\n----------------------------")
    print(f"Game: n = {n}, s = {s}")
    print("----------------------------")
    heaps = [0]*n
    player = 1
    print("Phase 1")
    while s > 0:
        heap = clever(heaps, s)
        heaps[heap] += 1
        s -= 1
        print(f"Player {player} places a stone in heap {heap+1}: {heaps}")
        player = 3-player
    print("\nPhase 2")
    print("Final heaps:", heaps)
    if play:
        while sum(heaps) > 0:
            heap, take = clever_nim(heaps)
            heaps[heap] -= take
            print(f"Player {player} removes {take} stone(s) from heap {heap+1}: {heaps}")
            if sum(heaps) == 0:
                print(f"Player {player} loses.")
                print(f"Player {3-player} wins.")
                return
            player = 3-player
    else:
        if nim_winning(heaps):
            print(f"Player {player} wins.")
        else:
            print(f"Player {3-player} wins.")


two_phase_nim(2, 3)
two_phase_nim(2, 4)
two_phase_nim(5, 4)
two_phase_nim(5, 3)

```

### Investigating the Nim sum

I then looked at the Nim sum to explain the results. The cases $(2,4)$, $(2,6)$ and $(2,5)$ are the examples discussed in **Attack**.


```python
# Calculate the Nim sum

def nim_sum(heaps):
    x = 0
    for h in heaps:
        x = x ^ h
    return x
# Rewrite nim_winning using the separate nim_sum function
def nim_winning(heaps):
    if max(heaps) <= 1:
        return sum(heaps) % 2 == 0
    return nim_sum(heaps) != 0
# Look at the cases discussed in Attack
for s in [2, 4, 5, 6]:
    two_phase_nim(2, s)
```

### Checking without the Nim sum

I wrote another recursive function to try every possible move in Phase 2. Unlike `nim_winning`, it does not use the Nim sum. This gave me a way to check my previous examples 

```python

# Check who can win without using the Nim sum

def check_winner(heaps):
    for i in range(len(heaps)):
        for take in range(1, heaps[i] + 1):
            new_heaps = heaps.copy()
            new_heaps[i] -= take
            # Taking the last stone loses
            if sum(new_heaps) == 0:
                continue
            if not check_winner(new_heaps):
                return True
    return False
```

### Comparing the results

I compared the two functions for the six configurations in the **Attack** table.

```python

# Compare the Nim sum method with the recursive method

positions = [
    [2, 2],
    [1, 1],
    [3, 2],
    [1, 1, 1],
    [1, 1, 1, 1],
    [3, 3]
]

for heaps in positions:
    print("\nHeaps:", heaps)
    print("Nim sum:", nim_sum(heaps))
    print("Recursive result:", check_winner(heaps))
    print("Nim sum prediction:", nim_winning(heaps))
    if check_winner(heaps) == nim_winning(heaps):
        print("Results agree")
    else:
        print("Results disagree")
```

# Playing a complete game

I then combined `clever`, which chooses placements, with `clever_nim`, which chooses removals. This lets me see the moves made in both phases.

```python
# Choose a good move in Phase 2

def clever_nim(heaps):
    for i in range(len(heaps)):
        for take in range(1, heaps[i] + 1):
            new_heaps = heaps.copy()
            new_heaps[i] -= take
            if sum(new_heaps) == 0:
                continue
            if not check_winner(new_heaps):
                return i, take
    # If there is no winning move, take one stone
    for i in range(len(heaps)):
        if heaps[i] > 0:
            return i, 1
two_phase_nim(2, 2, play=True)
two_phase_nim(2, 4, play=True)
two_phase_nim(3, 4, play=True)
two_phase_nim(2, 5, play=True)
```

## Review — Extension 1: Standard Nim

I changed Phase 2 so that taking the final stone wins. Phase 1 stayed the same. The only change in the code is the rule used when the placements finish.

```python

def normal_nim_winning(heaps):
    return nim_sum(heaps) != 0
def normal_winning(heaps, s):
    if s == 0:
        return normal_nim_winning(heaps)
    for i in range(len(heaps)):
        new_heaps = heaps.copy()
        new_heaps[i] += 1

        if not normal_winning(new_heaps, s - 1):
            return True
    return False
```

### Comparing the two rules

I compared the winner under the original last-stone-loses rule and standard Nim. This is the computation used in my first **Review** extension, where I noticed that the winners agree whenever $s>n$ in the cases tested.

```python

print("n  s  Misere  Standard")
for n in range(2, 5):
    for s in range(1, 7):
        heaps = [0] * n
        if winning(heaps, s):
            misere = "P1"
        else:
            misere = "P2"
        if normal_winning(heaps, s):
            standard = "P1"
        else:
            standard = "P2"
        print(n, s, misere, standard)
```

## Review — Extension 2: Removing at most two stones

I kept the last-stone-loses rule but allowed players to remove only one or two stones from a single heap. I used the same recursion as before, with a smaller range of possible removals.

```python

def restricted_nim_winning(heaps):
    for i in range(len(heaps)):
        for take in range(1, min(2, heaps[i]) + 1):
            new_heaps = heaps.copy()
            new_heaps[i] -= take
            if sum(new_heaps) == 0:
                continue
            if not restricted_nim_winning(new_heaps):
                return True
    return False

def restricted_winning(heaps, s):
    if s == 0:
        return restricted_nim_winning(heaps)
    for i in range(len(heaps)):
        new_heaps = heaps.copy()
        new_heaps[i] += 1
        if not restricted_winning(new_heaps, s - 1):
            return True
    return False
```

### Comparing restricted and unrestricted moves

I compared this variation with the original game. Some winners change, including $(n,s)=(3,4)$, which I discuss in **Review**. Here “Original” means the unrestricted last-stone-loses game.

```python

print("n  s  Original  Restricted")
for n in range(2, 5):
    for s in range(1, 7):
        heaps = [0] * n
        if winning(heaps, s):
            original = "P1"
        else:
            original = "P2"
        if restricted_winning(heaps, s):
            restricted = "P1"
        else:
            restricted = "P2"
        print(n, s, original, restricted)

```
